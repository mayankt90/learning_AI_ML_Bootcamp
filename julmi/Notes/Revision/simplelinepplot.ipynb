```python
from matplotlib import pyplot as plt
```

```python
years = [1950, 1960, 1970, 1980, 1990, 2000, 2010]
populations = [2.5, 3.0, 3.7, 4.4, 5.3, 6.1, 6.9]

plt.plot(years, populations, marker='o', linestyle='solid', color='green', label='Population Growth')
plt.title('World Population Over Time')
plt.xlabel('Year')
plt.ylabel('Population (in billions)')
plt.grid(True)
plt.show()
```
