```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
X = 6 * np.random.rand(100,1) - 3
Y = 0.5 * X**2 + 1.5 * X + np.random.randn(100,1)

plt.plot(X,Y,'b.')
plt.xlabel('X')
plt.ylabel('Y')
# plt.show()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
## Lets implement the simple linear regression model
from sklearn.linear_model import LinearRegression
lin_reg = LinearRegression()
```

```python
lin_reg.fit(X_train,Y_train)
```

```python
from sklearn.metrics import r2_score
Y_pred = lin_reg.predict(X_test)
print("R^2 score: ", r2_score(Y_test,Y_pred))
```

```python
# Lets plot the predicted values
plt.plot(X_train,lin_reg.predict(X_train),color='red',label='Linear Regression')
plt.scatter(X_train,Y_train,color='blue',label='Train Data')
plt.xlabel('X')
plt.ylabel('Y')
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=2,include_bias=True)
X_train_poly = poly.fit_transform(X_train)
X_test_poly = poly.transform(X_test)
```

```python
from sklearn.metrics import r2_score
poly_reg = LinearRegression()
poly_reg.fit(X_train_poly,Y_train)
Y_pred_poly = poly_reg.predict(X_test_poly)
print("R^2 score: ", r2_score(Y_test,Y_pred_poly))
```

```python
plt.scatter(X_train,poly_reg.predict(X_train_poly),color='red',label='Polynomial Regression')
plt.scatter(X_train,Y_train,color='blue',label='Train Data')
plt.xlabel('X')
plt.ylabel('Y')
```
